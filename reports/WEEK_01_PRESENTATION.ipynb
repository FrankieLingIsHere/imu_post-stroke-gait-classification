# Week 1 Progress Presentation

A concise presentation of the current gait-classification baseline, dataset roles, validation evidence, and remaining clinical-readiness gaps.

## Main message

The source-balanced Felius + Voisard baseline is a credible research prototype, but the current participant coverage does not support broad age-general or clinical-ready claims.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch
ROOT = Path.cwd()
if ROOT.name == 'reports': ROOT = ROOT.parent
metrics = pd.DataFrame({
    'Evaluation': ['Felius + Voisard pooled','Felius only','Voisard only','RevalExo frozen external'],
    'Participants': [284,163,121,17],
    'AUROC': [0.957,0.930,0.982,0.871],
    'Meaning': ['Internal pooled discrimination','Source-specific result','Source-specific result','Untouched external result']
})
display(metrics.style.format({'AUROC':'{:.3f}'}).hide(axis='index'))
```

## Baseline metrics

```python
fig, ax = plt.subplots(figsize=(10,5))
bars = ax.bar(metrics['Evaluation'], metrics['AUROC'], color=['#1b998b','#1976a3','#1976a3','#d99a00'])
ax.set_ylim(0,1); ax.set_ylabel('Participant-level AUROC'); ax.set_title('Frozen baseline performance')
ax.tick_params(axis='x', rotation=18)
for b,v in zip(bars,metrics['AUROC']): ax.text(b.get_x()+b.get_width()/2,v+0.025,f'{v:.3f}',ha='center',fontweight='bold')
fig.tight_layout(); fig.savefig(ROOT/'reports'/'baseline_metrics.png',dpi=180,bbox_inches='tight'); plt.show()
```

## Readiness conclusion

**Fulfilled:** participant-level labels, common preprocessing, disjoint validation, source-balanced training, internal/external discrimination, simpler-model comparison, and leakage controls.

**Still incomplete:** age-specific stroke validation, clinical-severity generalization, full calibration, and clinical deployment validation.

The model should be reported as a research prototype, not a clinical-ready system.
